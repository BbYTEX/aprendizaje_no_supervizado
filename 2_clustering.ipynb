```python
import pandas as pd

import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
df = pd.read_csv('./data/retailmax.csv')
```

```python
features = df[['Age', 'Annual Income (k$)', 'Spending Score (1-100)']]
```

```python
scaler = StandardScaler()
scaled_features = scaler.fit_transform(features)
```

```python
type(scaled_features)
```

```python
kmeans = KMeans(n_clusters=3, init='k-means++', max_iter=300, n_init=10, random_state=42)
clusters = kmeans.fit_predict(scaled_features)
```

```python
    # Función para calcular la Suma de Cuadrados Dentro del Cluster (WCSS)
def calcular_wcss(datos):
    wcss = []
    for n in range(1, 11):
        kmeans = KMeans(n_clusters=n, init='k-means++', max_iter=300, n_init=10, random_state=42)
        kmeans.fit(datos)
        wcss.append(kmeans.inertia_)
    return wcss

# Calcular el WCSS para diferentes números de clusters
wcss = calcular_wcss(scaled_features)

# Graficar el método del codo
plt.plot(range(1, 11), wcss)
plt.title('Método del codo')
plt.xlabel('Número de clusters')
plt.ylabel('WCSS')
plt.show()
```

```python
# Modelo final con cinco clusters
kmeans_final = KMeans(n_clusters=5, init='k-means++', max_iter=300, n_init=10, random_state=42)
df['Cluster'] = kmeans_final.fit_predict(scaled_features)

# Visualización 3D de los clusters
fig = plt.figure(figsize=(11, 8))
ax = fig.add_subplot(111, projection='3d')
scatter = ax.scatter(
    df['Annual Income (k$)'],
    df['Spending Score (1-100)'],
    df['Age'],
    c=df['Cluster'],
    cmap='Set1',
    s=55,
    alpha=0.85,
    edgecolors='white',
    linewidths=0.4
)
ax.set_xlabel('Ingreso anual (k$)', labelpad=10)
ax.set_ylabel('Puntuación de gasto (1-100)', labelpad=10)
ax.set_zlabel('Edad', labelpad=10)
ax.set_title('Segmentación de clientes en cinco clusters', pad=20, fontsize=16, fontweight='bold')
legend = ax.legend(*scatter.legend_elements(), title='Clusters', loc='upper left')
ax.add_artist(legend)
fig.tight_layout()
plt.show()

# Perfil de cada cluster
columnas_cluster = ['Age', 'Annual Income (k$)', 'Spending Score (1-100)', 'Cluster']
cluster_analysis = df[columnas_cluster].groupby('Cluster').mean().round(2)
cluster_sizes = df['Cluster'].value_counts().sort_index().rename('Clientes')
cluster_analysis.insert(0, 'Clientes', cluster_sizes)
print('Perfil de los clusters:')
display(cluster_analysis)

# Incorporar género sin modificar la columna original
features_extended = df[['Gender', 'Age', 'Annual Income (k$)', 'Spending Score (1-100)']].copy()
features_extended['Gender'] = features_extended['Gender'].map({'Male': 0, 'Female': 1})
scaled_extended_features = StandardScaler().fit_transform(features_extended)

kmeans_extended = KMeans(n_clusters=5, init='k-means++', max_iter=300, n_init=10, random_state=42)
df['Cluster Gender'] = kmeans_extended.fit_predict(scaled_extended_features)
features_extended['Cluster Gender'] = df['Cluster Gender']

extended_cluster_analysis = features_extended.groupby('Cluster Gender').agg(
    Clientes=('Cluster Gender', 'size'),
    Edad=('Age', 'mean'),
    Ingreso=('Annual Income (k$)', 'mean'),
    Gasto=('Spending Score (1-100)', 'mean'),
    Porcentaje_mujeres=('Gender', 'mean')
).round(2)
extended_cluster_analysis['Porcentaje_mujeres'] = (extended_cluster_analysis['Porcentaje_mujeres'] * 100).round(1)
print('Perfil de los clusters incorporando género:')
display(extended_cluster_analysis)

for cluster, perfil in extended_cluster_analysis.iterrows():
    print(f'Cluster {cluster}: {int(perfil["Clientes"])} clientes | '
          f'edad media {perfil["Edad"]:.1f} años | '
          f'ingreso medio {perfil["Ingreso"]:.1f} k$ | '
          f'gasto medio {perfil["Gasto"]:.1f} | '
          f'{perfil["Porcentaje_mujeres"]:.1f}% mujeres')
```
